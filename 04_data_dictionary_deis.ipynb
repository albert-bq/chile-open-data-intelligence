```python

import pandas as pd
from pathlib import Path
from IPython.display import display

DICT_PATH = Path(
    "/Volumes/workspace/codi_bronze/raw_files/"
    "extracted/2020/"
    "Diccionario BD egresos hospitalario.xlsx"
)

if not DICT_PATH.exists():
    raise FileNotFoundError(DICT_PATH)

# Identificar las hojas disponibles
excel = pd.ExcelFile(DICT_PATH, engine="openpyxl")

print("DICCIONARIO OFICIAL DEIS")
print("=" * 60)

for sheet in excel.sheet_names:
    df = pd.read_excel(
        excel,
        sheet_name=sheet,
        header=None,
        dtype=str
    )

    print(f"\nHOJA: {sheet}")
    print(f"FILAS: {len(df):,}")
    print(f"COLUMNAS: {len(df.columns)}")

    display(df.head(15))
```

```python

import pandas as pd

# El objeto excel ya fue creado anteriormente.
# Contiene las hojas del diccionario oficial.

print("=" * 70)
print("DICCIONARIO COMPLETO DE VARIABLES")
print("=" * 70)

df_variables = pd.read_excel(
    excel,
    sheet_name="Hoja1",
    header=0,
    dtype=str
)

for _, row in df_variables.iterrows():
    print(f"\nVARIABLE: {row.iloc[0]}")
    print(f"DESCRIPCIÓN: {row.iloc[1]}")

print("\n" + "=" * 70)
print("BÚSQUEDA DE VALORES ESPECIALES")
print("=" * 70)

terminos = [
    "*",
    "reserv",
    "confiden",
    "proteg",
    "sin información",
    "ignorado",
    "no informado",
    "no especificado"
]

for hoja in excel.sheet_names:

    df_hoja = pd.read_excel(
        excel,
        sheet_name=hoja,
        header=None,
        dtype=str
    ).fillna("")

    coincidencias = []

    for i, fila in df_hoja.iterrows():
        contenido = " | ".join(
            fila.astype(str).tolist()
        )

        if any(
            termino.lower() in contenido.lower()
            for termino in terminos
        ):
            coincidencias.append({
                "fila_excel": i + 1,
                "contenido": contenido[:1500]
            })

    print(f"\nHOJA: {hoja}")
    print("Coincidencias:", len(coincidencias))

    if coincidencias:
        display(pd.DataFrame(coincidencias).head(30))
```
